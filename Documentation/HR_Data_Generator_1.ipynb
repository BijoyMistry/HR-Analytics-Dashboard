# 🧠 Business Rules

The following rules are used while generating employee records.

These rules ensure every employee has realistic and logically connected information.

Examples:

- Older employees generally have more experience.
- Salary depends on department.
- Attrition is higher for employees working overtime.
- Managers usually have higher salaries.
- Freshers rarely have 15+ years of experience.

```python
# ==========================================================
# BUSINESS RULE FUNCTIONS
# ==========================================================

def generate_age():

    return random.randint(18,60)


def age_group(age):

    if age <= 25:
        return "18-25"

    elif age <= 35:
        return "26-35"

    elif age <= 45:
        return "36-45"

    elif age <= 55:
        return "46-55"

    else:
        return "55+"


def generate_experience(age):

    max_exp = max(age-21,0)

    return random.randint(0,max_exp)


def salary_slab(salary):

    if salary < 300000:
        return "0-3 LPA"

    elif salary < 600000:
        return "3-6 LPA"

    elif salary < 1000000:
        return "6-10 LPA"

    else:
        return "10+ LPA"
```

```python
# ==========================================================
# ATTRITION BUSINESS RULE
# ==========================================================

def generate_attrition(overtime, job_satisfaction):

    probability = 0.08

    if overtime == "Yes":
        probability += 0.15

    if job_satisfaction <= 2:
        probability += 0.12

    if random.random() < probability:
        return "Yes"

    return "No"
```

```python
# ==========================================================
# EMPLOYEE STATUS
# ==========================================================

def employee_status(attrition):

    if attrition == "Yes":
        return "Resigned"

    return "Active"
```

```python
# ==========================================================
# SALARY HIKE %
# ==========================================================

def salary_hike():

    return random.randint(5,20)
```

```python
# ==========================================================
# PERFORMANCE RATING
# ==========================================================

def performance():

    return random.choices(

        [1,2,3,4],

        weights=[8,22,45,25]

    )[0]
```

```python
# ==========================================================
# JOB SATISFACTION
# ==========================================================

def satisfaction():

    return random.choices(

        [1,2,3,4],

        weights=[15,25,35,25]

    )[0]
```

This section generates all employee records using the business rules defined earlier.

Every employee receives:

- Unique Employee Code
- Personal Information
- HR Information
- Salary Information
- Employment Details
- Performance Metrics

The generated dataset closely resembles a real enterprise HRMS.

```python
# ==========================================================
# GENERATE EMPLOYEE MASTER DATASET
# ==========================================================

!pip install Faker # Added to ensure Faker is available for this cell
TOTAL_EMPLOYEES = 2500 # Added to resolve NameError
import random # Added to resolve NameError

genders = [
    "Male",
    "Female"
] # Added to resolve NameError

from faker import Faker # Added to resolve NameError for 'fake'
fake = Faker("en_IN") # Added to resolve NameError for 'fake'

employees = []

for i in range(1, TOTAL_EMPLOYEES + 1):

    employee_code = f"HR{100001+i}"

    gender = random.choice(genders)

    if gender == "Male":
        employee_name = fake.name_male()
    else:
        employee_name = fake.name_female()

    age = generate_age()

    agegroup = age_group(age)

    experience = generate_experience(age)

    department = random.choice(departments)

    jobrole = random.choice(department_jobroles[department])

    salary = random.randint(
        salary_ranges[department][0],
        salary_ranges[department][1]
    )

    salarieslab = salary_slab(salary)

    travel = random.choice(business_travel)

    workmode = random.choice(work_mode)

    overtime_status = random.choice(overtime)

    marital = random.choice(marital_status)

    education = random.choice(education_levels)

    jobsat = satisfaction()

    envsat = random.choice(environment_satisfaction)

    worklife = random.choice(work_life_balance)

    perf = performance()

    hike = salary_hike()

    attrition = generate_attrition(
        overtime_status,
        jobsat
    )

    status = employee_status(attrition)

    joining_date = fake.date_between(
        start_date=START_DATE,
        end_date=END_DATE
    )

    employees.append({

        "EmployeeCode": employee_code,

        "EmployeeName": employee_name,

        "Gender": gender,

        "Age": age,

        "AgeGroup": agegroup,

        "Department": department,

        "JobRole": jobrole,

        "AnnualSalary": salary,

        "SalarySlab": salarieslab,

        "Experience": experience,

        "JoiningDate": joining_date,

        "BusinessTravel": travel,

        "WorkMode": workmode,

        "OverTime": overtime_status,

        "MaritalStatus": marital,

        "Education": education,

        "JobSatisfaction": jobsat,

        "EnvironmentSatisfaction": envsat,

        "WorkLifeBalance": worklife,

        "PerformanceRating": perf,

        "SalaryHikePercent": hike,

        "Attrition": attrition,

        "EmploymentStatus": status

    })

print("Employee Dataset Generated Successfully")
```

```python
# ==========================================================
# ATTRITION ENGINE
# ==========================================================

def calculate_attrition(employee):

    score = 0

    # Younger employees leave more often
    if employee["Age"] < 30:
        score += 2

    # Low satisfaction
    if employee["JobSatisfaction"] <= 2:
        score += 3

    # Poor work-life balance
    if employee["WorkLifeBalance"] <= 2:
        score += 2

    # Frequent travel
    if employee["BusinessTravel"] == "Travel_Frequently":
        score += 2

    # Overtime
    if employee["OverTime"] == "Yes":
        score += 2

    # Low experience
    if employee["YearsAtCompany"] <= 2:
        score += 2

    # Low salary
    if employee["SalarySlab"] == "0-3 LPA":
        score += 2

    probability = min(score / 30, 0.18)

    return "Yes" if random.random() < probability else "No"


print("Attrition Engine Ready")
```

```python
# ==========================================================
# BUSINESS LOGIC FUNCTIONS
# ==========================================================

# -------------------------------
# EXPERIENCE
# -------------------------------

def generate_experience(age):

    max_exp = max(age - 21, 0)

    return round(random.uniform(0, max_exp), 1)


# -------------------------------
# TENURE
# -------------------------------

def years_at_company(total_exp):

    return random.randint(0, max(0, int(total_exp)))


def years_current_role(years_company):

    return random.randint(0, max(0, years_company))


def years_since_promotion(years_company):

    return random.randint(0, max(0, years_company // 2))


def years_with_manager(years_company):

    return random.randint(0, max(0, years_company))


# -------------------------------
# SALARY
# -------------------------------

salary_ranges = {

    "Administrative Executive": (18000,30000),
    "HR Executive": (22000,35000),
    "Recruiter": (22000,32000),
    "Payroll Executive": (25000,38000),

    "Sales Executive": (22000,45000),
    "Sales Manager": (50000,90000),

    "Software Engineer": (35000,70000),
    "Senior Software Engineer": (65000,110000),
    "Data Analyst": (35000,65000),
    "Data Scientist": (70000,120000),
    "Network Engineer": (35000,70000),

    "Finance Executive": (30000,55000),
    "Accountant": (28000,50000),

    "Marketing Executive": (28000,50000),
    "Digital Marketing Specialist": (35000,65000),

    "Operations Executive": (28000,50000),
    "Operations Manager": (60000,100000),

    "Manager": (80000,140000)
}

def generate_salary(department, jobrole, experience):

    # Base salary mainly depends on experience
    if experience < 2:
        salary = random.randint(18000, 32000)

    elif experience < 5:
        salary = random.randint(32000, 50000)

    elif experience < 10:
        salary = random.randint(50000, 80000)

    elif experience < 15:
        salary = random.randint(80000, 120000)

    else:
        salary = random.randint(120000, 180000)

    # Department adjustment
    if department == "IT":
        salary += random.randint(5000, 15000)

    elif department == "Finance":
        salary += random.randint(5000, 12000)

    elif department == "Sales":
        salary += random.randint(3000, 10000)

    elif department == "Marketing":
        salary += random.randint(3000, 8000)

    elif department == "Operations":
        salary += random.randint(2000, 7000)

    elif department == "Human Resources":
        salary += random.randint(1000, 5000)

    # Manager roles
    if "Manager" in jobrole:
        salary += random.randint(30000, 60000)

    # Round to nearest ₹50
    salary = int(round(salary / 50) * 50)

    return salary

print("Business Logic Functions Ready")
```

```python
# ==========================================================
# EMPLOYEE GENERATOR
# ==========================================================

def generate_employee(emp_no):

    employee = {}

    # ======================================================
    # EMPLOYEE IDENTITY
    # ======================================================

    employee["EmployeeCode"] = f"HRA{240000 + emp_no:05d}"

    gender = random.choice(["Male","Female"])

    employee["Gender"] = gender

    employee["EmployeeName"] = generate_employee_name(gender)

    username = (
        employee["EmployeeName"]
        .lower()
        .replace(" ",".")
    )

    employee["OfficialEmail"] = username + "@hrhub.co.in"

    employee["Username"] = username

    # ======================================================
    # COMPANY DETAILS
    # ======================================================

    employee["EmployeeCount"] = 1

    employee["Over18"] = "Y"

    employee["StandardHours"] = 80

    employee["Company"] = COMPANY["CompanyName"]

    employee["Industry"] = COMPANY["Industry"]

    employee["Country"] = COMPANY["Country"]

# ======================================================
# DEMOGRAPHICS
# ======================================================

    age = generate_age()

    employee["Age"] = age

    employee["AgeGroup"] = age_group(age)

    employee["DateOfBirth"] = dob_from_age(age)

    employee["MaritalStatus"] = random.choices(
        ["Single", "Married", "Divorced"],
        weights=[42, 53, 5],
        k=1
)[0]

    education = random.choice(education_levels)

    employee["Education"] = education

    employee["EducationField"] = random.choice(
        education_fields[education]
)

    state = random.choice(states)

    employee["State"] = state

    employee["City"] = random.choice(
        cities[state]
)

    employee["PhoneNumber"] = generate_phone()

# ======================================================
# ORGANIZATION
# ======================================================

    department = random.choice(departments)

    employee["Department"] = department

    employee["JobRole"] = random.choice(
    department_roles[department]
)

    employee["BusinessTravel"] = random.choices(
        business_travel,
        weights=[60, 20, 20],
        k=1
)[0]

    employee["WorkMode"] = random.choices(
        work_modes,
        weights=[65, 25, 10],
        k=1
)[0]

# ======================================================
# EXPERIENCE
# ======================================================

    experience = generate_experience(employee["Age"])

    employee["TotalWorkingYears"] = experience

    employee["YearsAtCompany"] = years_at_company(experience)

    employee["YearsInCurrentRole"] = years_current_role(
        employee["YearsAtCompany"]
)

    employee["YearsSinceLastPromotion"] = years_since_promotion(
        employee["YearsAtCompany"]
)

    employee["YearsWithCurrManager"] = years_with_manager(
        employee["YearsAtCompany"]
)

    employee["HireDate"] = hire_date_from_experience(experience)

# ======================================================
# SALARY
# ======================================================

    salary = generate_salary(
        employee["Department"],
        employee["JobRole"],
        employee["TotalWorkingYears"]
)

    employee["MonthlyIncome"] = salary

    employee["SalarySlab"] = salary_slab(salary)

    employee["SalaryHikePercent"] = random.randint(5, 25)

    employee["StockOptionLevel"] = random.randint(0, 3)

# ======================================================
# PERFORMANCE & SATISFACTION
# ======================================================

    employee["PerformanceRating"] = random.choices(
        [1, 2, 3, 4],
        weights=[5, 15, 60, 20],
        k=1
)[0]

    employee["JobSatisfaction"] = random.randint(1, 4)

    employee["EnvironmentSatisfaction"] = random.randint(1, 4)

    employee["RelationshipSatisfaction"] = random.randint(1, 4)

    employee["WorkLifeBalance"] = random.randint(1, 4)

    employee["TrainingTimesLastYear"] = random.randint(0, 6)

    employee["NumCompaniesWorked"] = min(
        random.randint(0, 10),
        employee["TotalWorkingYears"]
)

# ======================================================
# PROMOTION
# ======================================================

    employee["PromotionLast5Years"] = random.choices(
        ["Yes", "No"],
        weights=[18, 82],
        k=1
)[0]

# ======================================================
# OVERTIME
# ======================================================

    employee["OverTime"] = random.choices(
        ["Yes", "No"],
        weights=[28, 72],
        k=1
)[0]

# ======================================================
# ATTRITION
# ======================================================

    employee["Attrition"] = calculate_attrition(employee)

    return employee

print("Employee Generator Created Successfully")
```
